```python
students = [
    {"name": "Rahul", "marks": 85},
    {"name": "Anjali", "marks": 72},
    {"name": "Kiran", "marks": 91},
    {"name": "Priya", "marks": 64},
    {"name": "Arjun", "marks": 78}
]

print(students)
```

```python
class DatasetIterator:

    def __init__(self, data):
        self.data = data
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):

        if self.index >= len(self.data):
            raise StopIteration

        item = self.data[self.index]

        self.index += 1

        return item
```

```python
dataset = DatasetIterator(students)

print(dataset)
```

```python
print(next(dataset))
```

```python
print(next(dataset))
```

```python
for student in dataset:
    print(student)
```

```python
dataset = DatasetIterator(students)

for student in dataset:

    if student["marks"] >= 40:
        print(
            student["name"],
            "→ Pass"
        )
```

```python
def create_discount(rate):

    def calculate_discount(price):
        return price - (price * rate / 100)

    return calculate_discount
```

```python
discount10 = create_discount(10)

print(discount10(1000))
```

```python
discount20 = create_discount(20)

print(discount20(1000))
```

```python
discount30 = create_discount(30)

print(discount30(1000))
```

```python
print("10% discount:", discount10(500))
print("20% discount:", discount20(500))
print("30% discount:", discount30(500))
```
