```python
salaries = np.array([35000, 48000, 52000, 67000, 45000, 75000, 58000, 42000, 90000, 51000])
print(salaries)
```

```python
total = np.sum(salaries)
print("Total Salary:", total)
```

```python
average = np.mean(salaries)
print("Average Salary:", average)
```

```python
maximum = np.max(salaries)
print("Maximum Salary:", maximum)
```

```python
minimum = np.min(salaries)
print("Minimum Salary:", minimum)
```

```python
above_50000 = salaries[salaries > 50000]
print("Salaries above 50000:", above_50000)
```

```python
count = np.sum(salaries > 50000)
print("Number of employees:", count)
```

```python
numbers = np.random.randint(1, 101, 20)
print(numbers)
```

```python
sorted_numbers = np.sort(numbers)
print("Sorted Numbers:", sorted_numbers)
```

```python
unique_numbers = np.unique(numbers)
print("Unique Numbers:", unique_numbers)
```

```python
standard_deviation = np.std(numbers)
print("Standard Deviation:", standard_deviation)
```

```python
numbers = np.arange(1, 101)
print(numbers)
```

```python
even_numbers = numbers[numbers % 2 == 0]
print("Even Numbers:", even_numbers)
```

```python
odd_numbers = numbers[numbers % 2 != 0]
print("Odd Numbers:", odd_numbers)
```

```python
divisible_by_5 = numbers[numbers % 5 == 0]
print("Numbers divisible by 5:", divisible_by_5)
```

```python
between_30_60 = numbers[(numbers >= 30) & (numbers <= 60)]
print("Numbers between 30 and 60:", between_30_60)
```
