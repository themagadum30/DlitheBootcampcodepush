```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("hospital_data.csv")
sns.boxplot(data=df, y="Treatment_Cost")
plt.title("Treatment Cost Distribution")
plt.show()
```

```python
df = pd.read_csv("customer_churn.csv")
sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)
plt.title("Monthly Charges by Churn")
plt.show()
```

```python
df = pd.read_csv("customer_churn.csv")
sns.countplot(
    data=df,
    x="Churn"
)
plt.title("Customer Churn Count")
plt.show()
```

```python
df = pd.read_csv("hospital_data.csv")
sns.countplot(
    data=df,
    x="Department"
)
plt.title("Patients by Department")
plt.xticks(rotation=45)

plt.show()
```

```python
df = pd.read_csv("hospital_data.csv")
sns.violinplot(
    data=df,
    x="Outcome",
    y="Treatment_Cost"
)
plt.title("Treatment Cost by Outcome")
plt.xticks(rotation=45)

plt.show()
```

```python
sns.violinplot(
    data=df,
    x="Gender",
    y="Age"
)
plt.title("Age Distribution by Gender")
plt.show()
```

```python
df = pd.read_csv("hospital_data.csv")
sns.regplot(
    data=df,
    x="Admission_Days",
    y="Treatment_Cost"
)
plt.title("Admission Days vs Treatment Cost")
plt.show()
```

```python
df = pd.read_csv("customer_churn.csv")
sns.regplot(
    data=df,
    x="Tenure",
    y="MonthlyCharges"
)
plt.title("Tenure vs Monthly Charges")
plt.show()
```
