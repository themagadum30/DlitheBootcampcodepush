```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Amit", "Sneha", "Rahul"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [55000, 45000, 70000, 65000, 48000],
    "Experience": [2, 5, 4, 6, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
high_salary = df[df["Salary"] > 50000]

print(high_salary)
```

```python
experienced = df[df["Experience"] > 3]

print(experienced)
```

```python
average_salary = df["Salary"].mean()

print("Average Salary:", average_salary)
```

```python
highest_paid = df[df["Salary"] == df["Salary"].max()]

print(highest_paid)
```

```python
import pandas as pd
import numpy as np

sales_data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Product": ["Laptop", "Mobile", "Laptop", "Headphones",
                "Mobile", "Laptop", "Headphones", "Mobile"],
    "Category": ["Electronics", "Electronics", "Electronics", "Accessories",
                 "Electronics", "Electronics", "Accessories", "Electronics"],
    "Quantity": [2, 3, 1, 5, 2, 1, 4, 3],
    "Price": [50000, 20000, 50000, 2000,
              20000, 50000, 2000, 20000],
    "Region": ["North", "South", "North", "East",
               "West", "South", "East", "West"]
}

df_sales = pd.DataFrame(sales_data)

print(df_sales)
```

```python
df_sales["Total_Amount"] = df_sales["Quantity"] * df_sales["Price"]

print(df_sales)
```

```python
total_revenue = df_sales["Total_Amount"].sum()

print("Total Revenue:", total_revenue)
```

```python
best_selling = df_sales.groupby("Product")["Quantity"].sum()

print(best_selling)
```

```python
best_product = best_selling.idxmax()

print("Best Selling Product:", best_product)
```

```python
print("Quantity Sold:", best_selling.max())
```

```python
category_revenue = df_sales.groupby("Category")["Total_Amount"].sum()

print(category_revenue)
```

```python
highest_category = category_revenue.idxmax()

print("Highest Revenue Category:", highest_category)
```

```python
region_sales = df_sales.groupby("Region")["Total_Amount"].sum()

print(region_sales)
```

```python
top_region = region_sales.idxmax()

print("Region with Highest Sales:", top_region)
```

```python
top_5_orders = df_sales.sort_values(
    by="Total_Amount",
    ascending=False
).head(5)

print(top_5_orders)
```

```python
print(df_sales.duplicated())
```

```python
print("Duplicate Records:", df_sales.duplicated().sum())
```

```python
df_sales = df_sales.drop_duplicates()

print(df_sales)
```

```python
print(df_sales.isnull().sum())
```

```python
df_sales["Quantity"] = df_sales["Quantity"].fillna(
    df_sales["Quantity"].mean()
)

df_sales["Price"] = df_sales["Price"].fillna(
    df_sales["Price"].mean()
)
```

```python
df_sales["Product"] = df_sales["Product"].fillna("Unknown")

df_sales["Category"] = df_sales["Category"].fillna("Unknown")

df_sales["Region"] = df_sales["Region"].fillna("Unknown")
```

```python
df_sales["Total_Amount"] = (
    df_sales["Quantity"] * df_sales["Price"]
)
```

```python
df_sales.to_csv("cleaned_sales.csv", index=False)
print("Cleaned dataset saved successfully!")
```
