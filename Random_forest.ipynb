```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import(accuracy_score,classification_report,confusion_matrix)
```

```python
data = {
    "Study_Hours": [2, 5, 1, 6, 8, 3, 7, 2, 9, 4,
                    1, 6, 8, 3, 7, 2, 10, 4, 5, 9],

    "Attendance": [60, 85, 50, 90, 95, 70, 92, 55, 98, 80,
                   45, 88, 94, 65, 90, 58, 99, 75, 82, 96],

    "Assignment_Marks": [40, 75, 30, 85, 90, 55, 88, 35, 95, 70,
                         25, 82, 92, 50, 85, 38, 98, 65, 72, 94],

    "Previous_Marks": [35, 70, 25, 80, 85, 50, 82, 30, 90, 65,
                       20, 78, 88, 45, 80, 32, 95, 60, 68, 91],

    "Internal_Marks": [38, 72, 28, 82, 88, 52, 85, 32, 92, 68,
                       22, 80, 90, 48, 82, 35, 96, 62, 70, 93],

    "Result": [
        "Fail", "Pass", "Fail", "Pass", "Pass",
        "Fail", "Pass", "Fail", "Pass", "Pass",
        "Fail", "Pass", "Pass", "Fail", "Pass",
        "Fail", "Pass", "Pass", "Pass", "Pass"
    ]
}

df = pd.DataFrame(data)

df
```

```python
x = df[
    [
    "Study_Hours",
    "Attendance",
    "Assignment_Marks",
    "Previous_Marks",
    "Internal_Marks"
]
]

y=df["Result"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42,stratify=y)
```

```python
model=RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print(y_pred)
```

```python
print("accuracy:",accuracy_score(y_test,y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
importance = pd.DataFrame({
    "Feature": x.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print(importance)
```

```python
new_student = [[
    6,    
    90,   
    85,   
    80,
    82    
]]

prediction = model.predict(new_student)

print("Prediction:", prediction[0])
```

```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import(accuracy_score,classification_report,confusion_matrix)
```

```python
data = {
    "Tenure": [2, 10, 24, 36, 5, 48, 12, 60, 3, 30,
               8, 40, 15, 55, 4, 20, 35, 50, 6, 45],

    "Monthly_Charges": [80, 60, 50, 45, 90, 40, 70, 35, 95, 55,
                        85, 42, 65, 38, 100, 58, 48, 40, 88, 44],

    "Support_Calls": [5, 2, 1, 0, 6, 0, 3, 0, 7, 1,
                      5, 0, 2, 0, 8, 2, 1, 0, 6, 1],

    "Total_Charges": [160, 600, 1200, 1620, 450, 1920, 840, 2100, 285,
                      1650, 680, 1680, 975, 2090, 400, 1160, 1680,
                      2000, 528, 1980],

    "Contract_Length": [1, 12, 24, 24, 1, 36, 12, 36, 1, 24,
                        1, 36, 12, 36, 1, 12, 24, 36, 1, 24],

    "Churn": [
        "Yes", "No", "No", "No", "Yes",
        "No", "Yes", "No", "Yes", "No",
        "Yes", "No", "Yes", "No", "Yes",
        "No", "No", "No", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

df
```

```python
X = df[
    [
        "Tenure",
        "Monthly_Charges",
        "Support_Calls",
        "Total_Charges",
        "Contract_Length"
    ]
]

y = df["Churn"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print(importance)
```

```python
new_customer = [[
    4,      
    90,     
    6,      
    360,    
    1      
]]

prediction = model.predict(new_customer)

print("Churn Prediction:", prediction[0])
```
